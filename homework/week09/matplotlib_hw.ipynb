```python
# Use only contiguous US states
state_map = contiguous_us.copy()

# Count observations that fall within each state
kudzu_join = gpd.sjoin(
    state_map[["STUSPS", "geometry"]],
    kudzu_points[["geometry"]],
    how="left",
    predicate="intersects"
)

pine_join = gpd.sjoin(
    state_map[["STUSPS", "geometry"]],
    pine_points[["geometry"]],
    how="left",
    predicate="intersects"
)

kudzu_counts = kudzu_join.groupby("STUSPS")["index_right"].count()
pine_counts = pine_join.groupby("STUSPS")["index_right"].count()

state_map["kudzu_count"] = (
    state_map["STUSPS"].map(kudzu_counts).fillna(0)
)

state_map["pine_count"] = (
    state_map["STUSPS"].map(pine_counts).fillna(0)
)
```

```python
schemes = {
    "Equal Interval": "equal_interval",
    "Quantile": "quantiles",
    "Natural Breaks (Jenks)": "natural_breaks",
    "Continuous": "continuous"
}

variables = [
    ("kudzu_count", "Kudzu"),
    ("pine_count", "Loblolly Pine")
]

for scheme_name, scheme in schemes.items():

    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    for ax, (column, species_name) in zip(axes, variables):

        zero = state_map[state_map[column] == 0]
        positive = state_map[state_map[column] > 0]

        zero.plot(
            ax=ax,
            color="lightgray",
            edgecolor="black",
            linewidth=0.4
        )

        if not positive.empty:
            plot_args = {
                "column": column,
                "cmap": "YlGnBu",
                "legend": True,
                "edgecolor": "black",
                "linewidth": 0.4,
                "ax": ax
            }

            if scheme == "equal_interval":
                plot_args.update(
                    scheme="equal_interval", k=5
                )

            elif scheme == "quantiles":
                plot_args.update(
                    scheme="quantiles",
                    k=min(5, positive[column].nunique())
                )

            elif scheme == "natural_breaks":
                plot_args.update(
                    scheme="NaturalBreaks",
                    k=min(5, positive[column].nunique())
                )

            positive.plot(**plot_args)

        ax.set_title(species_name)
        ax.set_axis_off()

    fig.suptitle(
        f"{scheme_name}: Species Observations by State",
        fontsize=14
    )

    plt.tight_layout()

    filename = (
        scheme_name.lower()
        .replace(" ", "_")
        .replace("(", "")
        .replace(")", "")
    )

    plt.savefig(
        f"{filename}_choropleth.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Use the state polygons and species counts from earlier code
bivariate = state_map.copy()

species = ["kudzu_count", "pine_count"]

# Fill states with no recorded observations with zero
bivariate[species] = bivariate[species].fillna(0)

# Classify each species:
# 0 = zero records
# 1 = positive count at or below the positive-count median
# 2 = count above the positive-count median

for name in species:
    values = bivariate[name]
    positive_values = values[values > 0]

    median = positive_values.median()

    bivariate[name + "_class"] = np.select(
        [values == 0, (values > 0) & (values <= median)],
        [0, 1],
        default=2
    )

    print(name, "positive-count median:", median)

# Rows represent increasing kudzu counts.
# Columns represent increasing loblolly pine counts.
palette = np.array([
    ["#eeeeee", "#c5b5d8", "#80619e"],
    ["#e6b899", "#b992a8", "#785b82"],
    ["#cf763e", "#a35e63", "#63384f"]
])

bivariate["color"] = [
    palette[kudzu_class, pine_class]
    for kudzu_class, pine_class in zip(
        bivariate["kudzu_count_class"],
        bivariate["pine_count_class"]
    )
]

# Draw the map
fig, ax = plt.subplots(figsize=(13, 8))

bivariate.plot(
    color=bivariate["color"],
    edgecolor="white",
    linewidth=0.5,
    ax=ax
)

ax.set_title(
    "Kudzu and Loblolly Pine Observations by State",
    fontsize=15
)
ax.set_axis_off()

# Create a 3 x 3 bivariate legend
legend_ax = ax.inset_axes([0.04, 0.04, 0.22, 0.27])

for kudzu_class in range(3):
    for pine_class in range(3):
        legend_ax.add_patch(
            Rectangle(
                (pine_class, kudzu_class), 1, 1,
                facecolor=palette[kudzu_class, pine_class],
                edgecolor="white"
            )
        )

legend_ax.set_xlim(0, 3)
legend_ax.set_ylim(0, 3)
legend_ax.set_aspect("equal")

legend_ax.set_xticks([0.5, 1.5, 2.5])
legend_ax.set_yticks([0.5, 1.5, 2.5])

legend_ax.set_xticklabels(
    ["Zero", "Lower", "Higher"], fontsize=8
)
legend_ax.set_yticklabels(
    ["Zero", "Lower", "Higher"], fontsize=8
)

legend_ax.set_xlabel("Loblolly pine records", fontsize=9)
legend_ax.set_ylabel("Kudzu records", fontsize=9)
legend_ax.tick_params(length=0)

plt.savefig(
    "bivariate_kudzu_loblolly_pine.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Choropleth maps can be great for visualizing species distributions, but the type of grouping/classifying utilized can skew how the data is perceived. Continuous color scales in choropleth maps makes it difficult for the viewer to match the color present on the map with the count value associated on the color ramp. How the data is classified into classes also affects how the data is visualized and a classification (quantile, equal interval, etc.) should be choosen based on the data's distribution.

```python
import mapclassify
from matplotlib.patches import Patch, Circle
from matplotlib.lines import Line2D

def pieMap(ax, contiguous_us, state_map):
    species = ["kudzu_count", "pine_count"]
    colors = ["forestgreen", "darkorange"]

    state_map[species] = state_map[species].fillna(0)
    state_map["total"] = state_map[species].sum(axis=1)

    positive = state_map.loc[state_map["total"] > 0].copy()

    # Three natural-breaks classes based on total observations
    if positive["total"].nunique() >= 3:
        classification = mapclassify.NaturalBreaks(
            positive["total"], k=3
        )
        positive["size_class"] = classification.yb
    else:
        # Handle datasets with fewer than three unique positive totals
        positive["size_class"] = pd.cut(
            positive["total"],
            bins=3,
            labels=False,
            duplicates="drop"
        ).fillna(0).astype(int)

    # Draw state polygons
    state_map.plot(
        ax=ax,
        facecolor="whitesmoke",
        edgecolor="gray",
        linewidth=0.5
    )

    # Preserve the map extent
    xlim, ylim = ax.get_xlim(), ax.get_ylim()

    # Pie radii in projected map meters, ordered small to large
    radii = [35000, 70000, 105000]

    for _, state in positive.iterrows():
        center = state.geometry.centroid

        # Skip states where one species has no records
        values = [state[species[0]], state[species[1]]]

        ax.pie(
            values,
            colors=colors,
            center=(center.x, center.y),
            radius=radii[int(state["size_class"])],
            startangle=90,
            wedgeprops={
                "edgecolor": "white",
                "linewidth": 0.5
            }
        )

    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    ax.set_aspect("equal")

    # Species legend
    species_legend = ax.legend(
        handles=[
            Patch(color=colors[0], label="Kudzu"),
            Patch(color=colors[1], label="Loblolly pine")
        ],
        title="Species",
        loc="lower left"
    )
    ax.add_artist(species_legend)

    # Size legend: proportional circles for the three total-count classes
    ax.figure.canvas.draw()

    points_per_meter = (
        ax.get_window_extent().width
        / (xlim[1] - xlim[0])
        * 72 / ax.figure.dpi
    )

    handles = []
    for class_id, radius in enumerate(radii):
        values = positive.loc[
            positive["size_class"] == class_id, "total"
        ]

        if not values.empty:
            handles.append(
                Line2D(
                    [], [],
                    linestyle="none",
                    marker="o",
                    markersize=2 * radius * points_per_meter,
                    markerfacecolor="lightgray",
                    markeredgecolor="gray",
                    label=f"{values.min():,.0f}–{values.max():,.0f}"
                )
            )

    ax.legend(
        handles=handles,
        title="Total records",
        loc="lower right",
        labelspacing=1.5
    )

    ax.set_title("Kudzu and Loblolly Pine Records by State")
    ax.set_axis_off()

    return ax
```

```python
# Project everything to a coordinate system measured in meters
states = contiguous_us.to_crs(epsg=5070)
kudzu_points = kudzu_points.to_crs(epsg=5070)
pine_points = pine_points.to_crs(epsg=5070)

# Combine state polygons to define the study area
us_boundary = states.geometry.union_all()

# Hexagon radius: 50 km
# Increase this value for larger hexagons; decrease for more detail.
radius = 50000

xmin, ymin, xmax, ymax = states.total_bounds

# Create a regular, flat-top hexagonal grid
dx = 1.5 * radius
dy = np.sqrt(3) * radius

hexagons = []
hex_id = []

row = 0
y = ymin

while y <= ymax + dy:
    x_offset = 0 if row % 2 == 0 else dx / 2
    x = xmin

    while x <= xmax + dx:
        cx = x + x_offset

        vertices = [
            (
                cx + radius * np.cos(np.radians(angle)),
                y + radius * np.sin(np.radians(angle))
            )
            for angle in range(0, 360, 60)
        ]

        polygon = Polygon(vertices)

        # Keep hexagons whose centers fall within the contiguous US
        if us_boundary.covers(polygon.centroid):
            hexagons.append(polygon)
            hex_id.append(len(hex_id))

        x += dx

    y += dy
    row += 1

hex_grid = gpd.GeoDataFrame(
    {"hex_id": hex_id, "geometry": hexagons},
    crs=states.crs
)

# Count the observations of each species within every hexagon
def count_points_in_hexagons(hex_grid, points, column_name):
    joined = gpd.sjoin(
        points[["geometry"]],
        hex_grid[["hex_id", "geometry"]],
        how="inner",
        predicate="within"
    )

    return joined.groupby("hex_id").size().rename(column_name)

kudzu_counts = count_points_in_hexagons(
    hex_grid, kudzu_points, "kudzu_count"
)
pine_counts = count_points_in_hexagons(
    hex_grid, pine_points, "pine_count"
)

hex_grid = hex_grid.join(kudzu_counts, on="hex_id")
hex_grid = hex_grid.join(pine_counts, on="hex_id")

hex_grid[["kudzu_count", "pine_count"]] = (
    hex_grid[["kudzu_count", "pine_count"]].fillna(0)
)

hex_grid["total"] = (
    hex_grid["kudzu_count"] + hex_grid["pine_count"]
)

# Display the grid itself, similar to your professor's first image
fig, ax = plt.subplots(figsize=(12, 8))
hex_grid.plot(ax=ax, color="steelblue", edgecolor="white", linewidth=0.3)
states.boundary.plot(ax=ax, color="gray", linewidth=0.3)
ax.set_axis_off()
ax.set_title("Hexagonal Grid for Species Distribution Maps")
plt.tight_layout()
plt.show()
```

```python
def classify_species(series):
    positive = series[series > 0]

    if positive.empty:
        return np.zeros(len(series), dtype=int)

    median = positive.median()

    return np.select(
        [
            series == 0,
            (series > 0) & (series <= median)
        ],
        [0, 1],
        default=2
    )

hex_grid["kudzu_class"] = classify_species(hex_grid["kudzu_count"])
hex_grid["pine_class"] = classify_species(hex_grid["pine_count"])

# Rows: kudzu zero/lower/higher
# Columns: pine zero/lower/higher
palette = np.array([
    ["#eeeeee", "#c5b5d8", "#80619e"],
    ["#e6b899", "#b992a8", "#785b82"],
    ["#cf763e", "#a35e63", "#63384f"]
])

hex_grid["bivariate_color"] = [
    palette[int(k), int(p)]
    for k, p in zip(
        hex_grid["kudzu_class"],
        hex_grid["pine_class"]
    )
]

fig, ax = plt.subplots(figsize=(12, 8))

hex_grid.plot(
    ax=ax,
    color=hex_grid["bivariate_color"],
    edgecolor="white",
    linewidth=0.3
)

states.boundary.plot(ax=ax, color="gray", linewidth=0.35)

# Create a 3 x 3 bivariate legend
legend_ax = ax.inset_axes([0.035, 0.035, 0.20, 0.25])

for k in range(3):
    for p in range(3):
        legend_ax.add_patch(
            Rectangle(
                (p, k), 1, 1,
                facecolor=palette[k, p],
                edgecolor="white"
            )
        )

legend_ax.set_xlim(0, 3)
legend_ax.set_ylim(0, 3)
legend_ax.set_aspect("equal")

legend_ax.set_xticks([0.5, 1.5, 2.5])
legend_ax.set_yticks([0.5, 1.5, 2.5])

legend_ax.set_xticklabels(["Zero", "Lower", "Higher"], fontsize=8)
legend_ax.set_yticklabels(["Zero", "Lower", "Higher"], fontsize=8)

legend_ax.set_xlabel("Loblolly pine records", fontsize=8)
legend_ax.set_ylabel("Kudzu records", fontsize=8)
legend_ax.tick_params(length=0)

ax.set_title("Kudzu and Loblolly Pine: Bivariate Hexagon Map")
ax.set_axis_off()

plt.tight_layout()
plt.savefig("hexagon_bivariate.png", dpi=300, bbox_inches="tight")
plt.show()
```
