I have decided to use gas plant locations, watershed boundaries, and county boundaries to identify gas plants located within surface water supply watersheds. 

Here are the sources of my data: 

watersheds: https://data-ncdenr.opendata.arcgis.com/datasets/fb32d3871a5640a986b72087c4121125_0/explore?location=35.894881%2C-80.389330%2C8

    - Provides geographic location of protected and critical areas of water supply as designated by NC DEQ Division of Water Resources

gas plants: https://www.nconemap.gov/datasets/a0722062315143a7980cebb79a204781_4/explore?location=35.265450%2C-79.408050%2C7

    - Provides geographic area of gas plants that participate in the MGP Assesment and Remediation Program. Gas sites are historical industrial sites that produced flammable gas from coal and oil until the 1970s. Gas sites within this program are notorious for contaminating groundwater and soil. 

nc counties: https://www.nconemap.gov/datasets/NCEM-GIS::north-carolina-state-and-county-boundary-polygons/explore?location=35.532328%2C-79.662726%2C7
    
    - Displays and contains data about North Carolina's state and county boundaries, also contains their area.

benthos: https://data-ncdenr.opendata.arcgis.com/datasets/ncdenr::nc-dwr-benthos-sample-results/about

    - Provides the most updated benthos monitoring data results and links to current and historic sampling analysis data sheets. Includes locations and sampling results for Macroinvertebrate sampling in NC's surface waters conducted by the NCDEQ Division of Water Resources - Water Sciences Section

water report: https://ordspub.epa.gov/ords/sfdw_rest/r/sfdw/sdwis_fed_reports_public/103
