```python
import time
import requests
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# getting the US census data

url_0 = "https://www2.census.gov/geo/tiger/GENZ2024/shp/" # url for census tiger data 
url_county = "cb_2024_us_county_500k.zip" # county level
url_state = "cb_2024_us_state_500k.zip" # state level 


### Code for county level data 

# counties = gpd.read_file(url_0+url_county)
# nc_counties = counties.loc[counties["STATEFP"] == "37"].copy()
# nc_counties.plot(edgecolor="black", facecolor="white")

## remove Alaska, Hawaii and those other islands which makes mapping difficult, but life much nicer.  
states = gpd.read_file(url_0+url_state)
contiguous_us = states.loc[
    ~states["STUSPS"].isin(["AK", "HI", "PR", "VI", "GU", "MP", "AS"])
]

contiguous_us = contiguous_us.to_crs(epsg=5070)
contiguous_us.plot(edgecolor="black", facecolor="white")
```

```python
kudzu = kudzu.dropna(
    subset=["decimalLatitude", "decimalLongitude"]
)

lob_pine = lob_pine.dropna(
    subset=["decimalLatitude", "decimalLongitude"]
)
```

```python
kudzu_points = gpd.GeoDataFrame(
    kudzu.copy(),
    geometry=gpd.points_from_xy( ## the method is calleed points_from_xy
        kudzu["decimalLongitude"], # x
        kudzu["decimalLatitude"] # y
        ),
    crs="EPSG:4326" # often the crs is in 4326 if it is a lat lon coordinate
    )

kudzu_points = kudzu_points.to_crs(epsg=5070)
```

```python
lob_points = gpd.GeoDataFrame(
    lob_pine.copy(),
    geometry=gpd.points_from_xy( ## the method is calleed points_from_xy
        lob_pine["decimalLongitude"], # x
        lob_pine["decimalLatitude"] # y
        ),
    crs="EPSG:4326" # often the crs is in 4326 if it is a lat lon coordinate
    )

lob_points = lob_points.to_crs(epsg=5070)
```

```python
# Convert coordinates to numbers
for df in [kudzu, lob_pine]:
    df["decimalLatitude"] = pd.to_numeric(
        df["decimalLatitude"], errors="coerce"
    )
    df["decimalLongitude"] = pd.to_numeric(
        df["decimalLongitude"], errors="coerce"
    )

    df.dropna(
        subset=["decimalLatitude", "decimalLongitude"],
        inplace=True
    )
```

```python
# Keep observations within the contiguous United States
kudzu = kudzu[
    kudzu["decimalLongitude"].between(-125, -66.5) &
    kudzu["decimalLatitude"].between(24, 50)
].copy()

lob_pine = lob_pine[
    lob_pine["decimalLongitude"].between(-125, -66.5) &
    lob_pine["decimalLatitude"].between(24, 50)
].copy()
```

```python
kudzu_points = gpd.GeoDataFrame(
    kudzu,
    geometry=gpd.points_from_xy(
        kudzu["decimalLongitude"],
        kudzu["decimalLatitude"]
    ),
    crs="EPSG:4326"
).to_crs(contiguous_us.crs)

pine_points = gpd.GeoDataFrame(
    lob_pine,
    geometry=gpd.points_from_xy(
        lob_pine["decimalLongitude"],
        lob_pine["decimalLatitude"]
    ),
    crs="EPSG:4326"
).to_crs(contiguous_us.crs)

# Plot
fig, ax = plt.subplots(figsize=(12, 8))

contiguous_us.plot(
    ax=ax, edgecolor="black",
    facecolor="white", linewidth=0.5
)

kudzu_points.plot(
    ax=ax, color="forestgreen",
    markersize=10, alpha=0.55,
    label="Kudzu"
)

pine_points.plot(
    ax=ax, color="darkorange",
    markersize=10, alpha=0.55,
    label="Loblolly pine"
)

ax.set_title(
    "Recorded Occurrences of Kudzu and Loblolly Pine"
)
ax.legend()
ax.set_axis_off()

plt.tight_layout()
plt.savefig("point_observation_map.png", dpi=300,
            bbox_inches="tight")
plt.show()
```

```python
# Use only contiguous US states
state_map = contiguous_us.copy()

# Count observations that fall within each state
kudzu_join = gpd.sjoin(
    state_map[["STUSPS", "geometry"]],
    kudzu_points[["geometry"]],
    how="left",
    predicate="intersects"
)

pine_join = gpd.sjoin(
    state_map[["STUSPS", "geometry"]],
    pine_points[["geometry"]],
    how="left",
    predicate="intersects"
)

kudzu_counts = kudzu_join.groupby("STUSPS")["index_right"].count()
pine_counts = pine_join.groupby("STUSPS")["index_right"].count()

state_map["kudzu_count"] = (
    state_map["STUSPS"].map(kudzu_counts).fillna(0)
)

state_map["pine_count"] = (
    state_map["STUSPS"].map(pine_counts).fillna(0)
)
```

```python
schemes = {
    "Equal Interval": "equal_interval",
    "Quantile": "quantiles",
    "Natural Breaks (Jenks)": "natural_breaks",
    "Continuous": "continuous"
}

variables = [
    ("kudzu_count", "Kudzu"),
    ("pine_count", "Loblolly Pine")
]

for scheme_name, scheme in schemes.items():

    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    for ax, (column, species_name) in zip(axes, variables):

        zero = state_map[state_map[column] == 0]
        positive = state_map[state_map[column] > 0]

        zero.plot(
            ax=ax,
            color="lightgray",
            edgecolor="black",
            linewidth=0.4
        )

        if not positive.empty:
            plot_args = {
                "column": column,
                "cmap": "YlGnBu",
                "legend": True,
                "edgecolor": "black",
                "linewidth": 0.4,
                "ax": ax
            }

            if scheme == "equal_interval":
                plot_args.update(
                    scheme="equal_interval", k=5
                )

            elif scheme == "quantiles":
                plot_args.update(
                    scheme="quantiles",
                    k=min(5, positive[column].nunique())
                )

            elif scheme == "natural_breaks":
                plot_args.update(
                    scheme="NaturalBreaks",
                    k=min(5, positive[column].nunique())
                )

            positive.plot(**plot_args)

        ax.set_title(species_name)
        ax.set_axis_off()

    fig.suptitle(
        f"{scheme_name}: Species Observations by State",
        fontsize=14
    )

    plt.tight_layout()

    filename = (
        scheme_name.lower()
        .replace(" ", "_")
        .replace("(", "")
        .replace(")", "")
    )

    plt.savefig(
        f"{filename}_choropleth.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Use the state polygons and species counts from earlier code
bivariate = state_map.copy()

species = ["kudzu_count", "pine_count"]

# Fill states with no recorded observations with zero
bivariate[species] = bivariate[species].fillna(0)

# Classify each species:
# 0 = zero records
# 1 = positive count at or below the positive-count median
# 2 = count above the positive-count median

for name in species:
    values = bivariate[name]
    positive_values = values[values > 0]

    median = positive_values.median()

    bivariate[name + "_class"] = np.select(
        [values == 0, (values > 0) & (values <= median)],
        [0, 1],
        default=2
    )

    print(name, "positive-count median:", median)

# Rows represent increasing kudzu counts.
# Columns represent increasing loblolly pine counts.
palette = np.array([
    ["#eeeeee", "#c5b5d8", "#80619e"],
    ["#e6b899", "#b992a8", "#785b82"],
    ["#cf763e", "#a35e63", "#63384f"]
])

bivariate["color"] = [
    palette[kudzu_class, pine_class]
    for kudzu_class, pine_class in zip(
        bivariate["kudzu_count_class"],
        bivariate["pine_count_class"]
    )
]

# Draw the map
fig, ax = plt.subplots(figsize=(13, 8))

bivariate.plot(
    color=bivariate["color"],
    edgecolor="white",
    linewidth=0.5,
    ax=ax
)

ax.set_title(
    "Kudzu and Loblolly Pine Observations by State",
    fontsize=15
)
ax.set_axis_off()

# Create a 3 x 3 bivariate legend
legend_ax = ax.inset_axes([0.04, 0.04, 0.22, 0.27])

for kudzu_class in range(3):
    for pine_class in range(3):
        legend_ax.add_patch(
            Rectangle(
                (pine_class, kudzu_class), 1, 1,
                facecolor=palette[kudzu_class, pine_class],
                edgecolor="white"
            )
        )

legend_ax.set_xlim(0, 3)
legend_ax.set_ylim(0, 3)
legend_ax.set_aspect("equal")

legend_ax.set_xticks([0.5, 1.5, 2.5])
legend_ax.set_yticks([0.5, 1.5, 2.5])

legend_ax.set_xticklabels(
    ["Zero", "Lower", "Higher"], fontsize=8
)
legend_ax.set_yticklabels(
    ["Zero", "Lower", "Higher"], fontsize=8
)

legend_ax.set_xlabel("Loblolly pine records", fontsize=9)
legend_ax.set_ylabel("Kudzu records", fontsize=9)
legend_ax.tick_params(length=0)

plt.savefig(
    "bivariate_kudzu_loblolly_pine.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
